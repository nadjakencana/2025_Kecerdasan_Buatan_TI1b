# Praktikum JobSheet 4 Kecerdasan Buatan

```python
import pandas as pd

#Mengubah DataSet menjadi Data Frame
df = pd.read_csv('./epl-goalScorer(20-21).csv')
```

```python
# Ambil 5 data pertama
print(df.head())
```

```python
# Tampilkan Jumlah Barsi dan Kolom
print(df.shape)
```

```python
# Tampilkan tipe data dari tiap kolom
print(df.dtypes)
```

```python
# Jumlah Goal tiap Pemain
print(df[['Player', 'Goals']])
```

```python
# DAtaframe berisi kolom tertentu
print(df[['Player', 'Goals', 'Position', 'Club', 'Yellow_Cards']])
```

```python
# Pemain Liverpool
print(df[df['Club'] == 'Liverpool'])
```

```python
# Pemain arsenal
print(df[df['Club'] == 'Arsenal'])
```

```python
# Top 10 kartu kuning
top_yellow = df.sort_values(by='Yellow_Cards', ascending=False)
print(top_yellow.head(10))
```

```python
# Top 10 Kartu merah
top_red = df.sort_values(by='Red_Cards', ascending=False)
print(top_red.head(10))
```

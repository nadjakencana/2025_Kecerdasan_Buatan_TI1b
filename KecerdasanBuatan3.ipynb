# Latihan

## Central Tendency

```python
#Mendapatkan nilai mean

import numpy as np
score = [90, 80, 70, 65, 110, 80, 83, 67, 94]
x = np.mean(score)
print(x)
```

```python
# Mendapatkan nilai median

import numpy as np
score = [90, 80, 70, 65, 110, 80, 83, 67, 94]
x = np.median(score)
print(x)
```

```python
#Mendapatkan nilai Maksimum

import numpy as np
score = [90, 80, 70, 65, 110, 80, 83, 67, 94]
nilai_maksimum = np.max(score)
print(nilai_maksimum)
print(x)
```

##Spread

```python
# Mendapatkan Nilai Range

import numpy as np

data = [2, 3, 7, 8, 9, 12, 15, 20, 50, 42]

minimum = np.min(data)
maximum = np.max(data)
x = maximum - minimum
print(x)
```

```python
# Mendapatkan nilai Percentile

import numpy as np

data = [90, 80, 70, 65, 110, 80, 83, 67, 94]

x = np.percentile(data, 90)
print(x)
```

```python
# Mendapatkan Nilai Quartile dan Interquartile Range

import numpy as np

data = [3, 4, 4, 5, 5, 6, 6, 7, 8, 8, 9, 10]

Q1 = np.median(data[:int(len(data)/2)])
Q3 = np.median(data[int(len(data)/2):])
IQR = Q3 - Q1

print(Q1, Q3, IQR)
```

```python
#mendapatkan nilai Variance dan Standard Deviation

import numpy as np

data = [86, 87, 88, 86, 87, 85, 86]

mean = np.mean(data)
var = np.var(data)
std = np.std(data)

print(mean, var, std)
```

## Praktikum 3

```python
# Urutan Data Hasil Ujian

data = [50, 54, 62, 50, 52, 59, 61, 63, 65, 10, 53, 63, 65, 50, 59, 62, 50, 51, 57, 60, 63, 65, 65, 53, 99]
data_sorted = sorted(data)

print("Data urut:", data_sorted)


#Cari Q1, Q3, dan IQR

import numpy as np

Q1 = np.percentile(data_sorted, 25)
Q3 = np.percentile(data_sorted, 75)
IQR = Q3 - Q1

print("Q1:", Q1)
print("Q3:", Q3)
print("IQR:", IQR)

#Hitung upper bound dan Lower Bound
lower_bound = Q1 - 1.5 * IQR
upper_bound = Q3 + 1.5 * IQR

print("Lower Bound:", lower_bound)
print("Upper Bound:", upper_bound)

#Tentukan Outlier
outliers = [x for x in data_sorted if x < lower_bound or x > upper_bound]

print("Outlier:", outliers)
```

## Praktikum 4

```python
import numpy as np

# Titik tengah setiap kelas
midpoints = [43, 48, 52.5, 58, 63, 68]

# Frekuensi
frequencies = [2, 3, 6, 4, 5, 2]

# Membuat daftar data berdasarkan frekuensi
data = []
for midpoint, freq in zip(midpoints, frequencies):
    data.extend([midpoint] * freq)

# Hitung standard deviation
std_dev = np.std(data)

print("Standar deviasi:", std_dev)
```
